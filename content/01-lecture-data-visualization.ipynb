## 3.2 Visualizing a single variable

### 3.2.1 A categorical variable: `employment`

## Step 1: start with the data

```r
ggplot(data = atus_college)
```

## Step 2: map a variable

```r
ggplot(data = atus_college, aes(x = employment))
```

## Step 3: add a geom

```r
ggplot(data = atus_college, aes(x = employment)) +
  geom_bar()
```

### 3.2.2 A numeric variable: `weekly_earnings`

```r
ggplot(data = atus_college, aes(x = weekly_earnings)) +
  geom_histogram()
```

## Changing the bin width

```r
ggplot(data = atus_college, aes(x = weekly_earnings)) +
  geom_histogram(binwidth = 20)
```

## Boxplots

A boxplot shows the five-number summary: minimum, first quartile,
median, third quartile, maximum. Points past 1.5 times the IQR
(interquartile range) are drawn as dots — possible outliers.

```r
ggplot(data = atus_college, aes(y = weekly_earnings)) +
  geom_boxplot()
```

## 3.3 Visualizing two variables

### 3.3.1 Two categorical variables

`employment` and `enrollment`. Map the second variable to `fill`.

```r
ggplot(
  data = atus_college,
  aes(x = employment, fill = enrollment)
) +
  geom_bar()
```

## Position: `"fill"`

Standardizes every bar to the same height, so you can compare
proportions.

```r
ggplot(
  data = atus_college,
  aes(x = employment, fill = enrollment)
) +
  geom_bar(position = "fill")
```

## Position: `"dodge"`

Puts the groups side by side instead of stacking them.

```r
ggplot(
  data = atus_college,
  aes(x = employment, fill = enrollment)
) +
  geom_bar(position = "dodge")
```

### 3.3.2 Two numeric variables

`time_alone` and `weekly_earnings`. Use `geom_point()` for a
scatterplot.

```r
ggplot(
  data = atus_college,
  aes(x = time_alone, y = weekly_earnings)
) +
  geom_point()
```

### 3.3.3 A numeric and a categorical variable

Side-by-side boxplots: `employment` (categorical, x) vs.
`weekly_earnings` (numeric, y).

```r
ggplot(
  data = atus_college,
  aes(x = employment, y = weekly_earnings)
) +
  geom_boxplot()
```

## 3.4 Visualizing more than two variables

Map extra variables to `color`, `shape`, or `size` to add more
dimensions to a scatterplot.

## Color by a categorical variable

```r
ggplot(
  data = atus_college,
  aes(x = time_alone, y = weekly_earnings, color = employment)
) +
  geom_point()
```

## Shape by a categorical variable

```r
ggplot(
  data = atus_college,
  aes(x = time_alone, y = weekly_earnings, shape = employment)
) +
  geom_point()
```

## Color *and* shape together

```r
ggplot(
  data = atus_college,
  aes(
    x = time_alone,
    y = weekly_earnings,
    color = employment,
    shape = employment
  )
) +
  geom_point()
```

## Size by a numeric variable

`size` works for numeric variables, like `work_time`.

```r
ggplot(
  data = atus_college,
  aes(
    x = time_alone,
    y = weekly_earnings,
    color = employment,
    size = work_time
  )
) +
  geom_point()
```
