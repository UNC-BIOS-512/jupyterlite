# Joining and pivoting

```r
library('dplyr')
library('ggplot2')
library('forcats')
library('lubridate')
```

```r
episodes = read.csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/refs/heads/main/data/2023/2023-11-28/drwho_episodes.csv')

episodes |> glimpse()
```

```r
writers = read.csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/refs/heads/main/data/2023/2023-11-28/drwho_writers.csv')

writers |> glimpse()
```

```r
options(repr.plot.width = 4)

writers |>
    mutate(writer = fct_lump(writer, n=10) |> 
               fct_infreq() |>
               fct_rev() |>
               fct_relevel('Other')
          ) |>
ggplot(aes(y = writer)) + geom_bar() + labs(y = '')
```

### Mutating joins

```r
left_join(
  x = country_rank,
  y = country_capital,
  by = "country"
)
```

```r
left_join(x = country_rank, y = country_capital, by = "country") |> 
  left_join(y = confederations, by = "confederation")
```

```r
right_join(
  x = country_rank,
  y = country_capital,
  by = "country"
)
```

```r
right_join(x = country_capital, y = country_rank, by = "country")
```

```r
full_join(
  x = country_rank,
  y = country_capital,
  by = "country"
)
```

```r
inner_join(x = country_rank, y = country_capital, by = "country")
```

### Filtering joins

```r
semi_join(x = country_rank, y = country_capital, by = "country")
```

```r
anti_join(x = country_rank, y = country_capital, by = "country")
```

### A summary of joins

| Join function | Rows from x | Columns from x | Rows from y | Columns from y |
| --- | --- | --- | --- | --- |
| `left_join()` | all | all | matched | all |
| `right_join()` | matched | all | all | all |
| `full_join()` | all | all | all | all |
| `inner_join()` | matched | all | matched | all |
| `semi_join()` | matched | all | none | none |
| `anti_join()` | unmatched | all | none | none |

### Pivot longer

```r
data(mx_us_wc_ranks)

mx_us_wc_ranks
```

```r
wc_long <- pivot_longer(
  data = mx_us_wc_ranks,
  cols = c("2010", "2014", "2018", "2022"),
  names_to = "year",
  values_to = "rank"
)
```

```r
wc_long
```

### Pivot wider

```r
pivot_wider(
  wc_long, 
  names_from = "year",
  values_from = "rank"
)
```
