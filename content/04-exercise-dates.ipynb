````r
library('dplyr')
library('ggplot2')
library('lubridate')
````

Our NYC 311 dataset has two full weeks of calls. I want to plot 📈 day of week (Monday-Sunday) versus number of calls for both weeks to see if there is a trend. The encodings will be **x-axis: weekday**, **y-axis: total calls**, and **color: week**.

First we load the data:

````r
nyc311 = read.csv('https://raw.githubusercontent.com/UNC-BIOS-512/jupyterlite/refs/heads/main/data/nyc311.csv')
nyc311 |> glimpse()
````

Now we need to make sure that the `created_date` column is stored as a date-time type. Use `mutate` and `ymd_hms` to convert `created_date` to a date-time.

<details>
<summary>Hint</summary>
    
```R
nyc311 = nyc311 |> mutate(created_date = ymd_hms(created_date))
```

</details>

**Ok! Now we just need to extract 🪨 the date info from the `created_date` column. Here's what you need to do:**

1. Create a new column called `week` using the `isoweek` function. ISO weeks run Monday-Sunday.
2. Create a new column called `wday` using the `wday` function. Use `label = TRUE` to get day names instead of numbers, and `week_start = 1` so that the week starts on Monday.
3. Once you have the columns set up, `count` by `week` and `wday`: `count(week, wday)`.

**Can you make the chart now?**

One gotcha: `wday` is a category (a factor), not a number. When the x-axis is a category, `geom_line` doesn't know which points to connect. Add `group = week` to tell it to draw one line per week.

Here is my solution ⬇︎

<details>
<summary>Solution</summary>

```R
nyc311 |>
    mutate(
        created_date = ymd_hms(created_date), 
        wday = wday(created_date, label = TRUE, week_start = 1),
        week = isoweek(created_date) |> as.factor()
    ) |> 
    count(week, wday) |> 
    ggplot(aes(x = wday, y = n, color = week, group = week)) +  # we need color and group here, why do you think? 🤔
    geom_line() + 
    geom_point() + 
    labs(x = 'Day', y = 'Number of 311 calls')
```
    
</details>
