# Scales and theme

## Scales

When we define an encoding we produce a scale.

```r
library('ggplot2')
library('palmerpenguins')

ggplot(penguins, aes(x = body_mass_g, y = flipper_length_mm, color = species)) + 
    geom_point(na.rm = T)
```

A scale maps values in your **data** (the *domain*) to values you can see on screen (the *range*) — a color, a position, a size, etc.

**Ordinal scale: species → color**



**Continuous scale: body mass → x position**




You can pick real colors for a scale from a palette.

[Link to ColorBrewer](https://colorbrewer2.org/#type=sequential&scheme=BuGn&n=3)

```r
ggplot(penguins, aes(x = body_mass_g, y = flipper_length_mm, color = species)) + 
    geom_point(na.rm = T) +
    scale_color_brewer(palette = 'Set2')
```

We can also transform scales.

```r
diamonds |> head()
```

```r
ggplot(diamonds, aes(x = price)) + 
    geom_histogram(bins = 50)
```

```r
ggplot(diamonds, aes(x = price)) + 
    geom_histogram(bins = 50) +
    scale_x_log10()
```

You can also change the words on an axis.

```r
ggplot(mpg, aes(displ, hwy)) + 
  geom_point(aes(colour = class)) + 
  scale_x_continuous(name = "A really awesome x axis label") +
  scale_y_continuous(name = "An amazingly great y axis label")
```

The name for a scale function follows a simple convention:

1. Start with `scale_`,
2. Add the name of the encoding (e.g. `x` or `color`),
3. Add the name of the type of scale you want to modify or add (`_continuous`, `_brewer`, etc.).

## Theme

Theme changes how a plot looks. It does not change the data.

```r
ggplot(mpg, aes(displ, hwy)) + 
    geom_point(aes(colour = class)) +
    theme(
        axis.title.x = element_text(color = 'red', face = 'italic'),  # change "title" to "text"
        panel.background = element_rect(fill = "lightblue"),
    )
```
