## Vectors of words

Words need quotation marks.

````r
names <- c("Menglin", "Gloria", "Robert")
names
````

## Building a data frame

````r
data.frame(friend = names, apples = n_apples)
````

## Store the data frame

````r
people <- data.frame(friend = names, apples = n_apples)
people
````

## Indexing: `data_frame[row, column]`

````r
people[2, 1]
````

## Whole rows and whole columns

````r
people[3, ]
````

````r
people[ ,2]
````

## Code style (Tidyverse)

- no space after a function name: `c(7, my_apples, 13)`
- spaces around operators: `<-`, `=`
- space *after* a comma, not before
- object names lower case, words joined by `_`
