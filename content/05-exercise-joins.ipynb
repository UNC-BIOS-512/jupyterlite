This will load some Friends data from the TidyTuesday project.

````r
friends = read.csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2020/2020-09-08/friends.csv') |> filter(!is.na(speaker))
emotions = read.csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2020/2020-09-08/friends_emotions.csv')
````

Use `glimpse` to make sense of these two tables.

By joining `friends` and `emotions`, we can count how many times a character's line (or "utterance") falls into the emotion categories.

Note that `emotions` only covers seasons 1-4, while `friends` has all 10 seasons. An `inner_join` keeps only the rows that match in *both* tables, so the later seasons are dropped silently. Check the seasons in your result to see this for yourself!

1. Use `inner_join` to combine `friends` and `emotions` by `c('season', 'episode', 'scene', 'utterance')`.
2. `count` by `season`, `speaker`, and `emotion`.
3. Save that in a new table called `char_emotion`.

<details>
<summary>Hint</summary>

```R
char_emotion = friends |> 
    inner_join(emotions, by = c('season', 'episode', 'scene', 'utterance')) |>
    count(season, speaker, emotion)
char_emotion |> glimpse()
```

</details>

Make a bar chart of the number of **Season 1 "Sad" utterances** for each character. Remember that for aggregated data we use `geom_col` and include both the `x` and `y` encoding!

<details>
<summary>Here is my solution 👀</summary>

```R
char_emotion |>
    filter(emotion == 'Sad', season == 1) |>
    mutate(speaker = speaker |> 
           fct_lump_n(n = 6, w = n) |>
           fct_reorder(n, .fun = sum) |>
           fct_relevel('Other')
          ) |>
ggplot(aes(y = speaker, x = n)) + geom_col() 
```
