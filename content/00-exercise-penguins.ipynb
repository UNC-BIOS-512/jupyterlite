## 2. Body mass vs. bill depth

Each point is one penguin. Colour shows the species.

R will print a warning about 2 removed rows. That is not a bug: two penguins in
this data set are missing measurements, so they cannot be drawn.

```r
ggplot(
    data = penguins,
    mapping = aes(x = body_mass_g, y = bill_depth_mm, color = species)
) + geom_point()
```
